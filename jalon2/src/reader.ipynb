```python
def dfs_spark():
    spark = SparkSession.builder.appName("ADLS Reader").getOrCreate()

    dfs = read_all_csvs_spark(spark, "raw")

    print("Loaded CSVs:", dfs.keys())

    # Show first CSV dynamically (avoids KeyError)
    first_key = list(dfs.keys())[0]
    dfs[first_key].show(5)

    return dfs
```
